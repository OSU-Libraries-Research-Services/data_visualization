### What is data literacy?

Data comes in many shapes, types, and forms. Data literacy “is the part of statistical literacy that involves training individuals to access, assess, manipulate, summarize, and present data.” __[(Calzada Prado and Marzel 2013)](https://doi.org/10.1515/libri-2013-0010)__

Data literacy involves:

Understanding data, including its context, who collected it and how;
Finding and/or obtaining data;
Reading, interpreting and evaluating data;
Managing data; and
Using data.

Data literacy concepts and competencies are interwoven throughout this course to help you master effective data visualization and storytelling. Beyond reading, interpreting, and evaluating data, we will cover data handling, data synthesis, ethical use of data, and using data. We will also cover best practices for visualizing data.
