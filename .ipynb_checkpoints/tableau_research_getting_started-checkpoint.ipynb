This section provides an introduction to the Tableau workspace and explores effective strategies for finding and critically assessing publicly available data. Section readings also offer guidance on understanding audience needs prior to data visualization and outline key principles for creating more impactful visual representations.

## Data skills | concepts
- Tableau
- Finding data
- Reading data

## Learning objectives
1. Apply the __[DRAMA Framework](https://www.librarianyarns.com/drama-source-eval#:~:text=DRAMA%20stands%20for%20Date%2C%20Relevance,in%20a%20one%20shot%20session.)__ to critically evaluate a dataset.
2. Navigate the Tableau start page, data source window, and workspace.
3. Connect, shape, and prepare data for analysis.
4. Change aggregation in a view.
