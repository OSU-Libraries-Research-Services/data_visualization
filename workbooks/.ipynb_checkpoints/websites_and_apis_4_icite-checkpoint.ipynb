# Measuring Impact: Exploring Scientific Influence with the iCite API
Lesson 4 of our Web Scraping and APIs for Digital Research workshop series explores how to measure the influence of scientific research using the iCite API from the National Institutes of Health. This session introduces the **Relative Citation Ratio (RCR)**, a powerful metric that benchmarks a publication’s citation performance against its peers.

Using real-world data from Ohio State University researchers, you’ll learn how to:
- Locate active NIH-funded projects using NIH RePORTER
- Retrieve publication data and RCR values using the iCite API
- Parse and store API responses using Python
- Interpret RCR scores to assess research impact
________________________________________
## What You’ll Learn:
- How to read and apply API documentation
- Construct and send API requests
- Extract and structure data from JSON responses
- Understand and interpret the Relative Citation Ratio (RCR)
________________________________________

The __[Relative Citation Ratio](https://doi.org/10.1371/journal.pbio.1002541)__ (RCR) evaluates the influence of a scientific article by analyzing its co-citation network. An RCR of **1.0** represents the median citation rate for articles in a given field. **Values above 1.0** indicate that an article is cited more frequently than average, suggesting greater impact or influence. In this lesson, we'll use the __[NIH Reporter](https://reporter.nih.gov/)__ to identify an active project at **The Ohio State University** with associated publications.  We will then retrieve the **RCR values** for thos publications using the __[iCite](https://icite.od.nih.gov/)__ API.

## Step 1: Locate an active NIH project
The __[NIH Reporter](https://reporter.nih.gov/)__ is a database of NIH-funded research projects that provides information about grants, contracts, publications, patents, and other outcomes resulting from NIH funding. 
- Go to the __[NIH Reporter](https://reporter.nih.gov/)__.
- Under **Fiscal Year**, select **Active Projects**.
- Under **Organization**, search for and select **Ohio State University**.

## Step 2: Browse active projects
- As of **September 5, 2025** there were **653 active projects** at The Ohio State University.
- Browse the list and click on a project that interests you.

## Step 3: Review project details
- Read the **project abstract** to understand its focus.
- To check for related publications:
    -  Click on **Publications** in the left-hand menu, or
    -  Scroll to the bottom of the project page. 

## Step 4: Export the publications
- If the publications are listed, use the **Export** option to download the list for further analysis.
