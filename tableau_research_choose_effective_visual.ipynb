# Working with data
Working with data requires **patience**, **practice**, and **curiosity**. If you're starting with a small dataset—just a few rows and columns—you can begin exploring it by creating simple charts and graphs. But keep in mind: **data is often messy** and may need to be cleaned or restructured before analysis.

Here are some common data preparation and enhancement tasks:

- **Standardizing** names, locations, and formats
- **Removing** unnecessary punctuation
- **Eliminating** duplicate rows (when appropriate)
- **Converting** text to consistent case (e.g., all lowercase)
- **Replacing** null values with zeros or other meaningful defaults
- **Standardizing** date and time formats
- **Splitting or unmerging** columns
- **Verifying and correcting** data types
- **Enhancing** your dataset by relating or joining it with a relevant external data source (e.g., adding demographic, geographic, or time-based context)

:::{caution} Null Values and Misclassified Data Types
Be especially careful with null values and checking data types. Nulls can skew calculations like averages or medians, and misclassified data types — such as dates stored as integers — can lead to errors in analysis. **Always verify that each column has the correct and consistent data type before building your visualizations.**
:::

## Relate or join data sources
**Relationships** combine data from multiple tables in a data source based on common fields **— without merging them into a single flat table.** This approach offers several benefits, including **preserving table-level detail** and helping to avoid issues like data duplication or incorrect aggregations.

To practice creating relationships and to better understand how they affect the level of detail in your visualizations, consider downloading Tableau’s __[Bookshop dataset](https://help.tableau.com/current/pro/desktop/en-us/bookshop_data.htm)__.

**Joins** in Tableau physically merge tables into a single flat table at the data source level, based on a defined join condition (e.g., inner, left, right, or full outer). While useful when both tables share the same level of detail, joins can lead to duplicated rows or inflated values if the join keys aren’t clean or consistent, so careful data preparation is essential.

# Choose an effective visual

It is a useful reference for selecting the right chart for the needs and expectations of your audience. This section will explore how to create basic Tableau charts in the **ranking**, **correlation**, and **change over time** categories.

:::

### Show Me

Another way to create a bar chart in Tableau is by using the **Show Me** ![](images/tableau_show_me.png "Decorative") button located in the upper right corner of the workspace. Before clicking **Show Me**, hold down the **Ctrl key** (or **Option key** on a Mac) while selecting one or more dimensions and/or measures. Tableau will then highlight the chart types that best match your selection. 

If you pre-select **artist**, **album_title**, and **Peak US**, for example, Tableau will automatically suggests horizontal bar charts as a suitable visualization.

![](images/tableau_show_me_barchart.png "Open show me tab with horizontal bars highlighted")

:::{admonition} Exercise 2. Bar Chart
:class: sidebar note
:icon: False

1. Open a new worksheet and rename the sheet **BarChart3**.
2. Drag **Release date** to columns and **rock_n_roll_studio_albums.csv (Count)** to rows.\
    **Note:** Tableau automatically creates a line chart, because **Release date** i+9s a **Date** field.
3. On the **Marks Card**, change the chart type from **Automatic** to **Bar**.

:::

## Change over time
or trends

### 📈 Line chart
Line charts are ideal for showing trends over time. When you place **Release date** on columns and **rock_n_r_stuio_albums.csv (Count)** on rows, Tableau automatically generates a line chart—recognizing the date field as discrete. 

Tableau categorizes fields as either **discrete** or **continuous** and uses pill color to denote the difference: **blue** for discrete and **green** for continuous. This distinction is especially important when working with dates, as they can be interpreted as continuous or discrete. This classification affects not only type of chart Tableau creates but also how you interact with and build your visualization. 

Let's walk through some examples to make this concept more concrete.

#### Discrete dates

1. Start a new worksheet and rename the sheet **LineChart1**.
2. Drag **Release date** to columns and **rock_n_roll_studio_albums.csv (Count)** to rows.
    - Notice that **Release date** is **blue**, indicating it's being treated as a **discrete** date.
    - The pill is labeled as **YEAR(Release date)** and is prefixed with a +, meaning it can be expanded.
3. Click the `+` icon next to YEAR to expand the date hierarchy:
    - Then click `+` **QUARTER(Release date)**.
    - Then click `+` **MONTH(Release date**.
    - Note that Tableau automatically parses discrete dates into their component parts: year, quarter, month, day.
4. Remove **YEAR**, **QUARTER**, and **DAY** from the view by dragging them off of the columns shelf.
5. Right-click while dragging **Release date** to the **Filters Shelf**.
    - This open the **Filter Field** window for dates.
    - Again, Tableau uses **blue** and **green** to indicate **discrete** versus **continuous** dates.
3. Select **Years** to filter by discrete year, then click **Next**.
4. Select **1960-1964**, then click **OK**.
5. Copy the **YEAR** filter to **Color** on the **Marks Card**
    - Hold down the Ctrl key (or Option key on a Mac) while moving **YEAR** on the Filters shelf to **Color** on the Marks card.

![](images/tableau_line_chart_1.png "Image showing steps 1-5  above completed")

{sup}`**Note:**The bulk of albums released in January is misleading. When gathering this data, if only a year was provided for release data, the month and day of the release date was automatically set to January 1.`

#### Continuous dates

Now let's see how the view changes when **Release date** is changed from a **discrete** to a **continuous** date.

1. Right click on **LineChart1** and select **Duplicate**.
2. Rename the duplicated sheet **LineChart2**.
3. Click the ▼ caret on the MONTH(Release date) pill.
4. Change the date from the discrete format (e.g. Month (May)) to the continuous format (e.g. Month (May 2015)).

At first glance, the chart might appear to be broken or disjointed. However, if you **hover over each individual line**, you'll see that Tableau is correctly plotting each month across all years on a continuous axis. The choppiness comes from the color encoding still being based on the **discrete** year. 

To smooth out the visualization and better highlight trends:

5. Remove the **YEAR(Release date) pill from **Color**.
6. IDrag a copy of **rock_n_roll_studio_albums.csv (Count)** to **Color**.

This applies a diverging color palette based on the count values, allowing you to visually emphasize peaks and valleys in the data—creating a more insightful and aesthetically pleasing trend chart.

![](images/tableau_line_chart_2.png "Image showing steps 1-6 above completed")


## Correlation 
or the relationship between two variables

### ![ ](images/tableau_scatterplot_icon.png) Scatterplot

## Text Table Alternatives
